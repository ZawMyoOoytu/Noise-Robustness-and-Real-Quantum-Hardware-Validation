```python
import numpy as np
import pandas as pd
from pathlib import Path

# Reproducibility
SEED = 42
rng = np.random.default_rng(SEED)

# Dataset size
N_SAMPLES = 1000

print("Dataset configuration")
print("=" * 40)
print("Samples:", N_SAMPLES)
print("Random seed:", SEED)
```

```python
# Generate synthetic 6G network intelligence features

sinr = rng.normal(15, 5, N_SAMPLES)
sinr = np.clip(sinr, -5, 30)

rsrp = rng.normal(-90, 10, N_SAMPLES)
rsrp = np.clip(rsrp, -120, -60)

latency = rng.gamma(shape=2.5, scale=8, size=N_SAMPLES)
latency = np.clip(latency, 2, 100)

packet_loss = rng.beta(1.5, 30, N_SAMPLES) * 10
packet_loss = np.clip(packet_loss, 0, 10)

throughput = rng.normal(250, 80, N_SAMPLES)
throughput = np.clip(throughput, 20, 500)

interference = rng.beta(2, 4, N_SAMPLES) * 100

resource_load = rng.beta(3, 2, N_SAMPLES) * 100

trust_score = rng.beta(8, 2, N_SAMPLES)

risk_score = 1 - trust_score

df = pd.DataFrame({
    "SINR": sinr,
    "RSRP": rsrp,
    "Latency": latency,
    "Packet_Loss": packet_loss,
    "Throughput": throughput,
    "Interference": interference,
    "Resource_Load": resource_load,
    "Trust_Score": trust_score,
    "Risk_Score": risk_score
})

print("Dataset shape:", df.shape)
print()
print(df.head())
```

```python
# Build a composite network-risk score

risk_signal = (
    0.35 * (1 - df["Trust_Score"]) +
    0.20 * (df["Risk_Score"]) +
    0.15 * (df["Latency"] / 100) +
    0.10 * (df["Packet_Loss"] / 10) +
    0.10 * (df["Interference"] / 100) +
    0.10 * (df["Resource_Load"] / 100)
)

df["Composite_Risk"] = risk_signal

# Binary classification target
threshold = df["Composite_Risk"].median()

df["Label"] = (
    df["Composite_Risk"] >= threshold
).astype(int)

print("Classification threshold:", threshold)
print()
print("Class distribution:")
print(df["Label"].value_counts().sort_index())
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

print("scikit-learn import: OK")
```

```python
PROJECT_ROOT = Path(r"C:\Users\heart\quantum_6g_ai")

DATASET_PATH = PROJECT_ROOT / "results" / "6g_trust_dataset.csv"

df = pd.read_csv(DATASET_PATH)

FEATURE_COLUMNS = [
    "SINR",
    "RSRP",
    "Latency",
    "Packet_Loss",
    "Throughput",
    "Interference",
    "Resource_Load",
    "Trust_Score",
    "Risk_Score"
]

TARGET_COLUMN = "Label"

X = df[FEATURE_COLUMNS].copy()
y = df[TARGET_COLUMN].copy()

print("Dataset shape:", df.shape)
print("Feature matrix:", X.shape)
print("Target vector :", y.shape)

print("\nFeatures:")
for i, feature in enumerate(FEATURE_COLUMNS, 1):
    print(f"{i:2}. {feature}")

print("\nTarget distribution:")
print(y.value_counts().sort_index())
```

```python
SEED = 42

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("\nTraining class distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts().sort_index())
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled testing shape :", X_test_scaled.shape)

print("\nTraining feature means:")
print(np.round(X_train_scaled.mean(axis=0), 4))

print("\nTraining feature std:")
print(np.round(X_train_scaled.std(axis=0), 4))
```

```python
preprocessed_train = pd.DataFrame(
    X_train_scaled,
    columns=FEATURE_COLUMNS
)

preprocessed_train["Label"] = y_train.to_numpy()

preprocessed_test = pd.DataFrame(
    X_test_scaled,
    columns=FEATURE_COLUMNS
)

preprocessed_test["Label"] = y_test.to_numpy()

TRAIN_PATH = PROJECT_ROOT / "results" / "train_scaled.csv"
TEST_PATH = PROJECT_ROOT / "results" / "test_scaled.csv"

preprocessed_train.to_csv(TRAIN_PATH, index=False)
preprocessed_test.to_csv(TEST_PATH, index=False)

print("Preprocessed datasets saved successfully:")
print(TRAIN_PATH)
print(TEST_PATH)

print("\nTrain shape:", preprocessed_train.shape)
print("Test shape :", preprocessed_test.shape)
```
